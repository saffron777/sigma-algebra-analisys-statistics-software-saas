## Relación entre precio y unidades vendidas

```
fig, ax = plt.subplots()
ax.scatter(ventas["precio"], ventas["unidades"], alpha=0.5, s=18)
slope, intercept = np.polyfit(ventas["precio"], ventas["unidades"], 1)
xs = np.linspace(ventas["precio"].min(), ventas["precio"].max(), 50)
ax.plot(xs, intercept + slope * xs, color="crimson")
ax.set_xlabel("Precio"); ax.set_ylabel("Unidades"); ax.set_title(f"Pendiente = {slope:.2f}")
plt.show()
```

## Regresión múltiple con statsmodels

```
import statsmodels.api as sm
X = sm.add_constant(ventas[["precio", "publicidad", "competidores"]])
modelo = sm.OLS(ventas["unidades"], X).fit()
print(modelo.summary())
```

```
online = ventas.loc[ventas["canal"] == "Online", "unidades"]
tienda = ventas.loc[ventas["canal"] == "Tienda", "unidades"]
stats.ttest_ind(online, tienda, equal_var=False)
```
